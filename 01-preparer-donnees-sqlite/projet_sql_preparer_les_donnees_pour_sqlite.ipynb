## Filtrer les données

```python
df['Year'].value_counts()
```

```python
# On garde seulement les 4 premiers caractères de la colonne 'Year'
```

```python
df['Year'] = df['Year'].str[0:4]
```

```python
# On vérifie le type de la colonne 'Year' 
```

```python
df['Year'].dtypes
```

```python
# Conversion de la colonne 'Year' de texte (string) vers un entier (int)
```

```python
df['Year'] = df['Year'].astype(int)
```

```python
# Vérifier à nouveau que la colonne 'Year' est bien un nombre entier
```

```python
df['Year'].dtypes
```

```python
# Filtrer et garder les années aprés 2000 de la colonne 'Year'
later_than_2000 = df[df['Year'] > 2000]
```

```python
# Affichage de quelques valeurs
later_than_2000.head()
```

```python
# Regarder combien de valeurs ont a désormais sur notre dataframe
later_than_2000.shape 
```

```python
# Actor -- Leading Role
# Actor -- Supporting Role
# Actress -- Leading Role
# Actress -- Supporting Role

# Création d'une liste qui contient ces 4 catégories
award_categories = ['Actor -- Leading Role', 'Actor -- Supporting Role', 
                    'Actress -- Leading Role', 'Actress -- Supporting Role']
```

```python
# Filtrer et garder uniquement les lignes qui contiennent ces 4 catégories avec la méthode 'ISIN' sur le dataset later_than_2000
nominations = later_than_2000[later_than_2000['Category'].isin(award_categories)].copy()
```

```python
# Vérifier le nombre de lignes et de colonnes
nominations.shape
```

```python
# Afficher quelques lignes (pour vérifier)
nominations.head()
```

## Nettoyer les colonnes Unnamed et Won

```python
# Création d'un dictionnaire de remplacement afin de remplacer les NO = 0 et Yes = 1
replacements = {"NO":0, "YES":1}
```

```python
nominations['Won?'] = nominations['Won?'].map(replacements)
```

```python
nominations["Won"] = nominations["Won?"]
```

```python
nominations.head()
```

```python
# Supprimer les colonnes inutiles
drop_cols = ["Won?", "Unnamed: 5", "Unnamed: 6", "Unnamed: 7", "Unnamed: 8", "Unnamed: 9", "Unnamed: 10"]
```

```python
# Supprimant les colonnes de 'drop_cols'
final_nominations = nominations.drop(drop_cols, axis=1)
```

```python
# Vérifier que tout est Ok 
final_nominations.head()
```

```python
# On peut voir désormais que la colonne 'WON?' et les autres colonnes 'Unnamed' ont été supprimés
```

## Nettoyage de la colonne Additionnal Info

```python
# On va séparer le nom du film, ainsi que l'acteur à partir de notre tableau final_nominations
# FILM {'PERSONNAGE'}

# Ensuite on va créer un tableau séparant les deux infos
# Biutiful {'Uxbal'}
# The Social Network --> Mark Zuckerberg
```

```python
additional_info_one = final_nominations['Additional Info'].str.rstrip("'}")
```

```python
additional_info_one.head()
```

```python
additional_info_two = additional_info_one.str.split(" {'", expand=True)
```

```python
additional_info_two.head()
```

```python
# Enregistrer dans une nouvelle variable = les films 
movie_names = additional_info_two[0]
```

```python
movie_names.head()
```

```python
# Enregistrer dans une nouvelle variable = les personnages 
characters = additional_info_two[1]
```

```python
characters.head()
```

```python
# Ajouter ses deux colonnes dans notre tableau 'final_nominations'
final_nominations['Movie'] = movie_names
final_nominations['Character'] = characters
```

```python
# Afficher le tableau 'final_nominations'
final_nominations.head()
```

```python
# Supprimer la colonne 'Additional Info' car on a splité les infos 
final_nominations.drop(['Additional Info'], axis=1, inplace=True)
final_nominations.head()
```

## Exporter vers SQLite sur DB Browser

```python
# méthode : to_sql(name, connexion)
```

```python
final_nominations.dtypes
```

```python
# Créer une nouvelle base de données et s'y connecter
connexion = sqlite3.connect('nominations.db')
```

```python
final_nominations.to_sql('nominations', connexion, index=False)
```
